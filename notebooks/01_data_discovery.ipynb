```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../data/raw")

sales_path = DATA_DIR / "sales_train_validation.csv"
calendar_path = DATA_DIR / "calendar.csv"
prices_path = DATA_DIR / "sell_prices.csv"

print("Sales file exists:", sales_path.exists())
print("Calendar file exists:", calendar_path.exists())
print("Prices file exists:", prices_path.exists())
```

```python
sales = pd.read_csv(sales_path)
calendar = pd.read_csv(calendar_path)
prices = pd.read_csv(prices_path)

print("Sales shape:", sales.shape)
print("Calendar shape:", calendar.shape)
print("Prices shape:", prices.shape)
```

```python
sales_sample = sales.head(100).copy()

print(sales_sample.shape)
```

```python
id_columns = [
    "id",
    "item_id",
    "dept_id",
    "cat_id",
    "store_id",
    "state_id"
]

day_columns = [
    col for col in sales_sample.columns
    if col.startswith("d_")
]

sales_long = sales_sample.melt(
    id_vars=id_columns,
    value_vars=day_columns,
    var_name="day",
    value_name="units_sold"
)
```

```python
sales_long.head(10)
```

```python
print("Number of day columns:", len(day_columns))
print("Expected rows:", len(sales_sample) * len(day_columns))
print("Actual rows:", len(sales_long))
```

```python
calendar["date"] = pd.to_datetime(calendar["date"])

sales_long = sales_long.merge(
    calendar[["d", "date", "wm_yr_wk"]],
    left_on="day",
    right_on="d",
    how="left"
)

sales_long.drop(columns=["d"], inplace=True)

sales_long.head()
```

```python
print("Rows:", sales_long.shape[0])
print("Missing dates:", sales_long["date"].isna().sum())
print("Missing week numbers:", sales_long["wm_yr_wk"].isna().sum())
```

```python
prices.head(10)
```

```python
sales_long = sales_long.merge(
    prices,
    on=["item_id", "store_id", "wm_yr_wk"],
    how="left"
)

sales_long.head()
```

```python
print("Shape:", sales_long.shape)

print("\nMissing values:")
print(sales_long.isna().sum())

print("\nData types:")
sales_long.info()
```

```python
missing_price_sales = sales_long[
    sales_long["sell_price"].isna()
]

print("Total missing-price records:",
      len(missing_price_sales))

print("Records with sales > 0:",
      (missing_price_sales["units_sold"] > 0).sum())

print("Records with sales = 0:",
      (missing_price_sales["units_sold"] == 0).sum())
```

```python
contradictions = sales_long[
    sales_long["sell_price"].isna() &
    (sales_long["units_sold"] > 0)
]

contradictions[
    ["item_id", "store_id", "date", "units_sold", "wm_yr_wk"]
].head(20)
```

```python
product_demand = (
    sales_long
    .groupby("item_id")["units_sold"]
    .sum()
    .sort_values(ascending=False)
)

product_demand.head(10)
```

```python
print("Total units sold:", sales_long["units_sold"].sum())
print("Number of products:", sales_long["item_id"].nunique())
print("Average units sold per product:",
      product_demand.mean())
```

```python
zero_sales = (
    sales_long
    .groupby("item_id")["units_sold"]
    .apply(lambda x: (x == 0).sum())
    .sort_values(ascending=False)
)

zero_sales.head(10)
```

```python

# Identifier / metadata columns
id_columns = [
    "id", "item_id", "dept_id",
    "cat_id", "store_id", "state_id"
]

# Daily sales columns
day_columns = [
    col for col in sales_full.columns
    if col.startswith("d_")
]

print("Metadata columns:", len(id_columns))
print("Daily sales columns:", len(day_columns))

print("First 5 days:", day_columns[:5])
print("Last 5 days:", day_columns[-5:])

print("Total historical days:", len(day_columns))
```

```python

# Total units sold across all 1913 days
category_sales = (
    sales_full.groupby("cat_id")[day_columns]
    .sum()
    .sum(axis=1)
    .sort_values(ascending=False)
)

print(category_sales)
```
